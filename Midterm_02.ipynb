## Context

When signing up for online sites, you are often asked to create a password that satisfies rules.  
These include, but are not limited to:
- Has a minimum number of characters.   
- Must have a mix of upper and lower case letters.  
- Must include at least one number.  
- Must include at least special character (e.g. "!" or "$").    

The reason for this is security: the more complex your password, the harder it is for malicious agents to guess your password through brute force or social engineering.

The *[Password Game](https://neal.fun/password-game/)* is this idea taken in the direction of absurdity.  
A player supplies a password, and the game checks if it satisfies the current set of rules.  
Assuming they pass the current set, the game then adds a new rule and the user must update the password to pass the expanded set of rules.  
The difficulty of the game is both in the difficulty of some of the rules, as well as the complexity of multiple overlapping rules.

## Problem 4

```python
def rule_4(pw: str) -> bool:
    """
    Rule to implement for problem 4.

    Includes the following conditions:
        - Contains only 3 types of characters: pipe-character,
          lowercase-L , and upppercase-I
        - The number of 'I', '|', and 'l' characters is equally balanced

    So a typical password looks like a 'barcode' such as:
        lI|
        ||llII
        IlI||llI|

    In this case, an empty string does count as valid.

    Parameters
    ----------
    pw : str
        Password to check for conformity

    Returns
    -------
    out : bool
        Does the password pass all the conditions of this rule

    """
    # Handle empty string case as valid
    if not pw:
        return True

    count_i = 0
    count_pipe = 0
    count_l = 0

    allowed_chars = {'I', '|', 'l'}

    for char in pw:
        if char not in allowed_chars:
            return False
        if char == 'I':
            count_i += 1
        elif char == '|':
            count_pipe += 1
        elif char == 'l':
            count_l += 1

    # Check if the counts are equally balanced
    return count_i == count_pipe == count_l
```

## Problem 5

```python
import pandas as pd
```

```python
def rule_5(pw: str) -> bool:
    """
    Rule to implement for problem 5.

    Includes the following conditions:
        - Must contain exactly 5 emoticons and nothing else.
        - Valid emoticons listed in external file.
        - No requirement that the emoticons be unique.

    Parameters
    ----------
    pw : str
        Password to check for conformity

    Returns
    -------
    out : bool
        Does the password pass all the conditions of this rule

    See Also
    --------
    data/emoticons.csv : List of emoticons and corresponding name.
                         Entries are newline-separated, comma-delimited.
                         Contains only 22 emoticons.
                         Strictly necessary this time.
    """
    emoticons_list = []
    try:
        # Read emoticons from the specified CSV file
        emoticons_df = pd.read_csv('/content/emoticons.csv', header=None, names=['emoticon', 'name'])
        emoticons_list = emoticons_df['emoticon'].tolist()
    except FileNotFoundError:
        # If the file is not found, the emoticon list remains empty.
        pass

    if not emoticons_list:
        # If no emoticons were loaded, it's impossible to satisfy the rule.
        return False

    # Sort emoticons by length in descending order to prioritize longer matches
    # This is important for cases where one emoticon might be a prefix of another (e.g., :) and :))
    emoticons_list.sort(key=len, reverse=True)

    emoticon_count = 0
    current_index = 0
    pw_len = len(pw)

    while current_index < pw_len:
        found_match = False
        for emoticon in emoticons_list:
            if pw.startswith(emoticon, current_index):
                emoticon_count += 1
                current_index += len(emoticon)
                found_match = True
                break  # Move to the next part of the password string
        if not found_match:
            # If no emoticon matches at the current index, the password contains invalid characters
            return False

    # The password must contain exactly 5 emoticons and entirely consist of them.
    return emoticon_count == 5
```

## Problem 6

```python
import pandas as pd
import re
```

```python
def rule_6(pw: str) -> bool:
    """
    Rule to implement for problem 6.

    Includes the following conditions:
        - The only allowed characters are ('X', 'V', 'I', '.')
        - The characters, when treated as Roman numerals, must
          MULTIPLY to 300.
        - Different numerals must be separated by at least 1 '.'

    Parameters
    ----------
    pw : str
        Password to check for conformity

    Returns
    -------
    bool
        Does the password pass all the conditions of this rule

    See Also
    --------
    data/Roman.csv : List of numbers and corresponding numeral.
                     Entries are newline-separated, comma-delimited.
                     Sorted numerically, up to 39 (largest possible).
                     Not strictly necessary, but should save time.

    """
    roman_map = {}
    try:
        # Read Roman numerals from the specified CSV file
        # Corrected: swap 'numeral' and 'value' names to match CSV structure (value, numeral)
        roman_df = pd.read_csv('/content/Roman.csv', header=None, names=['value', 'numeral'])
        # Filter numerals to include only those composed of 'X', 'V', 'I'
        for _, row in roman_df.iterrows():
            numeral = row['numeral'] # This is now the string (e.g., 'I', 'II', 'X')
            value = row['value']     # This is now the integer (e.g., 1, 2, 10)
            if all(char in {'X', 'V', 'I'} for char in numeral):
                roman_map[numeral] = value
    except FileNotFoundError:
        return False  # Cannot proceed without the Roman.csv file

    if not roman_map:
        return False  # No valid Roman numerals loaded after filtering

    # 1. Handle empty string case
    if not pw:
        return False  # An empty string cannot multiply to 300

    # 2. Validate overall password characters
    allowed_chars_set = {'X', 'V', 'I', '.'}
    if not all(char in allowed_chars_set for char in pw):
        return False

    # 3. Validate dot placement: no leading/trailing dots, no multiple adjacent dots
    if pw.startswith('.') or pw.endswith('.') or '..' in pw:
        return False

    # 4. Split and parse segments
    segments = pw.split('.')

    current_product = 1
    for segment in segments:
        if segment not in roman_map:
            return False  # Segment is not a valid Roman numeral from our filtered list
        current_product *= roman_map[segment]

    # 5. Final check
    return current_product == 300
```
