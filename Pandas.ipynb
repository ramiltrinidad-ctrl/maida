```python
np.random.seed(1337)
data = pd.DataFrame({'user_id': ['001', '002', '023'],
                     'milk': [0, 1, 2],
                     'cheese': [0, 0, 1],
                     'diaper': [1, 0, 1],
                     'beer': [1, 1, 0]}).set_index('user_id')
data.columns.name = 'product'
data
```

```python
def range_value(x):
  return x.max() - x.min()

display(range_value(updated_data['beer']))
```
