# Problem 6 [10 pts.]

Return a **COPY** of a dataframe, where all the `NaN` values in a column have been replaced by some float value `N`.  

May NOT exceed 1 line of working code (pandas one-liner).  
Empty lines and formatting (e.g. for hanging indents) are excluded.

```python
def column_fillna(
    data: pd.DataFrame = cotter,
    feature: str = 'Titration pH',
    N: float | np.float64 = 8.2
) -> pd.DataFrame:
    """
    Return a **COPY** of a dataframe, where all the `NaN` values in a column
    have been replaced by some float value `N`.

    May NOT exceed 1 line of working code (pandas one-liner).
    Empty lines and formatting (e.g. for hanging indents) are excluded.

    Parameters
    ----------
    data : pd.DataFrame
        Dataframe to clean

    feature : str
        Name of column in dataframe to clean (no need for guardrails for this)

    N: float or np.float64
        Fill value to use

    Returns
    -------
    pd.DataFrame
        COPY of original dataframe with NaNs in column filled
    """
    return data.copy().fillna({feature: N})
```

# Problem 7 [10 pts.]

Take a dataframe and return a pandas series indexed by a grouper feature, whose entries are the median over all entries in that group in the original dataframe.

May NOT exceed 1 line of working code (pandas one-liner).  
Empty lines and formatting (e.g. for hanging indents) are excluded.

```python
def median_of_group(
    data: pd.DataFrame = cotter,
    grouper: str = 'Brew',
    feature: str = 'Dose'
) -> pd.Series:
    """
    Take a dataframe and return a pandas series indexed by a grouper feature,
    whose entries are the median over all entries in that group in the
    original dataframe.

    May NOT exceed 1 line of working code (pandas one-liner).
    Empty lines and formatting (e.g. for hanging indents) are excluded.

    Parameters
    ----------
    data : pd.DataFrame
        Dataframe to process

    grouper : str
        Name of column in original dataframe to set groups; will become index

    feature : str
        Name of column in original dataframe to take median over. Clearly
        needs to correspond to a column with numerical entries, but again do
        not bother putting guardrails.

    N: float or np.float64
        Fill value to use

    Returns
    -------
    pd.Series
        Pandas series with median value of feature-values over each
        grouper-value in original dataframe.
    """
    return data.groupby(grouper)[feature].median()
```

# Problem 8 [10 pts.]

Take a dataframe and return a **COPY** of it with a new column that includes the sum of values between a set of column names.
The new column is always on the far-right, and there are no other changes compared to the original dataframe.

May NOT exceed 3 lines of working code.  
Empty lines and formatting (e.g. for hanging indents) are excluded.

```python
def sum_columns(
    data: pd.DataFrame = cotter,
    start_col: str = 'Tea.floral',
    end_col: str = 'Rubber',
    col_name: str = 'Sum'
) -> pd.DataFrame:
    """"
    Take a dataframe and return a COPY of it with a new column that includes
    the sum of values between a set of column names. The new column is always
    on the far-right, and there are no other changes compared to the original
    dataframe.

    May NOT exceed 3 lines of working code. Empty lines and formatting (e.g.
    for hanging indents) are excluded.

    Parameters
    ----------
    data : pd.DataFrame
        Dataframe to process

    start_col : str
        Name of first column to add as in slicing

    end_col : str
         Name of last column to add as in slicing

    col_name : str
        Name of newly-added column

    Returns
    -------
    pd.DataFrame
        COPY of original dataframe with sum of values between the set of column
        names appended to the right.
    """
    df_copy = data.copy()
    df_copy[col_name] = df_copy.loc[:, start_col:end_col].sum(axis=1)
    return df_copy
```

# Problem 9 [10 pts.]

Take a dataframe, and return a new dataframe indexed by a grouper feature, whose columns contain summary statistics for a list of features in hierarchical column index with the order of features preserved.  
The statistics per feature include (also in order):
1. Mean
1. Standard Error of the Mean
1. Standard Deviation
1. Skew

May NOT exceed 2 line of working code.  
Empty lines and formatting (e.g. for hanging indents) are excluded.

```python
def summary_stats(
    data: pd.DataFrame = cotter,
    grouper: str = 'Brew',
    features: list[str] = ['Liking',
                           'Acidity',
                           'Flavor.intensity',
                           'Mouthfeel',
                           'Temp']
) -> pd.DataFrame:
    """
    Take a dataframe, and return a dataframe indexed by a grouper feature, and
    whose columns contain summary statistics for a list of features in
    hierarchical column index, with the order of features preserved. The
    statistics per feature include (also in order):

        1. Mean
        2. Standard Error of the Mean
        3. Standard Deviation
        4. Skew

    May NOT exceed 2 line of working code. Empty lines and formatting (e.g.
    for hanging indents) are excluded.

    Parameters
    ----------
    data : pd.DataFrame
        Dataframe to process

    grouper : str
        Name of column in original dataframe to set groups; will become index

    features : list[str]
        List of features to create summary statistics for. Need to be numerical
        for intended statistics to work (no need for guardrails in code).

    Returns
    -------
    pd.DataFrame
        Pandas dataframe with summary statistics of selected features

    Notes
    -----
    Only columns are hierarchical, rows are simply indexed
    """
    return data.groupby(grouper)[features].agg(['mean', 'sem', 'std', 'skew'])
```

# Problem 10 [10 pts.]

Take the `cotter` dataset loaded above and return a **COPY** of the dataframe.  
In this copy, convert the `Temp.x`, `TDS.x`, and `PE.x` columns to the numerical equivalent SPECIFICALLY by reading the `Brew` values per index, and transferring the values over.  
The updated columns should also be converted to a numeric datatype; no other columns should be touched.

Brew values are '-' separated, and follow the same order as the columns.  
So `Brew` = '87-1.0-16' reads as `TMP.x` = $87$, `TDS.x` = $1.0$, `PE.x` = $16$  

May NOT exceed 5 line of working code.  
Empty lines and formatting (e.g. for hanging indents) are excluded.

```python
def format_columns(data: pd.DataFrame = cotter) -> pd.DataFrame:
    """
    Take the `cotter` dataset loaded above and return a COPY of the dataframe.
    In this copy, convert the Temp.x, TDS.x, and PE.x columns to the numerical
    equivalent SPECIFICALLY by reading the Brew` values per index, and
    transferring the values over. The updated columns should also be converted
    to a numeric datatype; no other columns should be touched.

    Brew values are '-' separated, and follow the same order as the columns. So
    Brew = '87-1.0-16' reads as TMP.x = 87, TDS.x = 1.0, PE.x = 16

    May NOT exceed 5 line of working code. Empty lines and formatting (e.g. for
    hanging indents) are excluded.

    Parameters
    ----------
    data : pd.DataFrame
        Dataframe to process (should just be the cotter dataset)

    Returns
    -------
    pd.DataFrame
        COPY of the cotter dataframe with the Temp.x, TDS.x, and PE.x columns
        updated to numerical values and datatype
    """
    df_copy = data.copy()
    df_copy[['Temp.x', 'TDS.x', 'PE.x']] = (
        df_copy['Brew'].str.split('-', expand=True).astype(float)
    )
    return df_copy
```
