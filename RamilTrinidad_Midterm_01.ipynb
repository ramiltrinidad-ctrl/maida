## Context

When signing up for online sites, you are often asked to create a password that satisfies rules.  
These include, but are not limited to:
- Has a minimum number of characters.   
- Must have a mix of upper and lower case letters.  
- Must include at least one number.  
- Must include at least special character (e.g. "!" or "$").    

The reason for this is security: the more complex your password, the harder it is for malicious agents to guess your password through brute force or social engineering.

The *[Password Game](https://neal.fun/password-game/)* is this idea taken in the direction of absurdity.  
A player supplies a password, and the game checks if it satisfies the current set of rules.  
Assuming they pass the current set, the game then adds a new rule and the user must update the password to pass the expanded set of rules.  
The difficulty of the game is both in the difficulty of some of the rules, as well as the complexity of multiple overlapping rules.

## Problem 1

```python
def rule_1(pw: str) -> bool:
    """
    Rule to implement for problem 1.

    Includes the following conditions:
        - Password must have length equal to a multiple of 10.
        - Password must contain characters to spell 'kartoffelkopf'

    Only care about having the characters in order. So there can be any
    number of characters around or between desired ones. For example:

         'oquw3478 K-A -R+T !OFF^ E=$ Lg K7OP.F sad5'

    would be a valid password (word is in uppercase).
    This rule should be case-INsensitive.

    Parameters
    ----------
    pw : str
        Password to check for conformity

    Returns
    -------
    out : bool
        Does the password pass all the conditions of this rule

    """
    # Condition 1: Password must have length equal to a multiple of 10.
    is_length_multiple_of_10 = len(pw) % 10 == 0

    # Condition 2: Password must contain characters to spell 'kartoffelkopf'
    # in order, case-insensitive.
    target_word = "kartoffelkopf"
    pw_lower = pw.lower()
    target_idx = 0

    for char_pw in pw_lower:
        if target_idx < len(target_word) and char_pw == target_word[target_idx]:
            target_idx += 1

    has_kartoffelkopf_subsequence = target_idx == len(target_word)

    # Both conditions must be met
    return is_length_multiple_of_10 and has_kartoffelkopf_subsequence
```

## Problem 2

```python
def rule_2(pw: str) -> bool:
    """
    Rule to implement for problem 2.

    Includes the following conditions:
        - Must contain only letters and numbers (no space/special char.)
        - If the letters are converted to numbers according to
          alphebetical position, the total sum of the password must be
          part of the Fibonacci sequence.

    For example, 'abcdef999' has sum 1+2+3+4+5+6+9+9+9 = 48

    The Fibonacci seqeunce, F_n, is defined by the recursive relation
        F_0 = 0
        F_1 = 1
        F_n = F_{n-1} + F_{n-2}
    so the first 9 Fibonacci numbers are (0, 1, 1, 2, 3, 5, 8, 13, 21).

    Parameters
    ----------
    pw : str
        Password to check for conformity

    Returns
    -------
    out : bool
        Does the password pass all the conditions of this rule

    See Also
    --------
    data/A1Z26.csv : List of abbreviations, name, atomic number.
                     Entries are newline-separated, tab-delimited.
                     Sorted alphabetically.
                     Not strictly necessary, but should save time.

    """
    total_sum = 0

    # Condition 1: Must contain only letters and numbers
    for char in pw:
        if not char.isalnum():
            return False

        if char.isalpha():
            # Convert letter to its alphabetical position (a=1, b=2, ...)
            # Use .lower() to handle both upper and lower case letters
            total_sum += (ord(char.lower()) - ord('a') + 1)
        elif char.isnumeric():
            # Convert digit to integer
            total_sum += int(char)

    # Condition 2: Total sum must be part of the Fibonacci sequence
    fib_numbers = {0, 1}  # F_0 and F_1
    a, b = 0, 1

    # Generate Fibonacci numbers until 'b' exceeds 'total_sum'
    # This ensures that if 'total_sum' itself is a Fibonacci number,
    # it's included.
    # The loop condition handles cases where total_sum is 0, 1, or other
    # small values correctly.
    while b <= total_sum:
        next_fib = a + b
        fib_numbers.add(next_fib)
        a = b
        b = next_fib

    return total_sum in fib_numbers
```

## Problem 3

```python
def rule_3(pw: str) -> bool:
    """
    Rule to implement for problem 3.

    Includes the following conditions:
        - Password must only contain 2-letter abbreviations for elements
        - The SUM of atomic numbers must be exactly 200

    In this case, all elements must be 2-character abbreviations
    following the periodic table. Elements that are normally abbreviated
    by 1-character should be converted to 2-character by having a space
    on the right. For example hydrogen is normally 'H', becomes 'H '.

    Parameters
    ----------
    pw : str
        Password to check for conformity

    Returns
    -------
    out : bool
        Does the password pass all the conditions of this rule

    See Also
    --------
    data/PT.txt : List of abbreviations, name, atomic number.
                  Entries are newline-separated, tab-delimited.
                  Sorted alphabetically by name (not abbreviation).
                  Not strictly necessary, but should save time.

    """
    # Check if the password length is a multiple of 2
    if len(pw) % 2 != 0:
        return False

    # Load element data from PT.txt
    element_atomic_numbers = {}
    try:
        with open('data/PT.txt', 'r') as f:
            for line in f:
                parts = line.strip().split('\t')
                if len(parts) == 3:
                    abbreviation = parts[0]
                    atomic_number = int(parts[2])
                    # Ensure 2-character abbreviation for lookup
                    if len(abbreviation) == 1:
                        key = abbreviation + ' '
                        element_atomic_numbers[key] = atomic_number
                    else:
                        element_atomic_numbers[abbreviation] = atomic_number
    except FileNotFoundError:
        # If the file is not found, we cannot validate, so the rule fails
        return False

    total_atomic_sum = 0
    # Iterate through the password in 2-character chunks
    for i in range(0, len(pw), 2):
        element_abbr = pw[i:i+2]
        if element_abbr not in element_atomic_numbers:
            return False  # Not a valid element abbreviation
        total_atomic_sum += element_atomic_numbers[element_abbr]

    return total_atomic_sum == 200
```
